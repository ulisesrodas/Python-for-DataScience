# <a id='1.'>I. `Pandas`</a>

## <a id='1.1.'>1.1. Definición y Características</a>

**DEFINICIÓN:**

* `Pandas` es una biblioteca de Python utilizada para trabajar con conjuntos de datos: vamos a llamarla cada vez que trabajamos con DataFrames.
* Es la librería más importante de aprender para limpiar, transforar y agregar datos.
* Como es externa, necesitamos importarla para trabajar en nuestro quehacer diario mediante la palabra clave `import`.

```python
import pandas as pd
import numpy as np
```

* Ahora, el paquete de `Pandas` está referido como `pd` en lugar de Pandas.

**CARACTERÍSTICAS:**

* La primera estructura de datos de `Pandas` es una Serie, que es un arreglo unidimensional que puede contener cualquier tipo de datos, de manera similar a un ndarray.
* Tiene un índice (index) que le asigna una etiqueta a cada entrada. Por lo general, el índice se utiliza para etiquetar los datos.
* Contiene información sobre una sola característica (feature) de los datos.
* Un `Pandas Dataframe` son muchas Series.

## <a id='1.2.'>1.2. Generación de `Series`<a>

### De `lists` a `Series`:

```python
# Recordemos una lista de Python, vista la clase pasada:
list1 = [0.25, 0.5, 0.75, 1.0]
list1
```

```python
# Se puede decir que una Serie es una columna:
pd.Series([1, 5, 8, 9, 10])
```

```python
list2 =  [1, 1.5, 2, 2.5, 3]
list2
```

```python
# Convertimos la lista en una Serie de Pandas:
data = pd.Series(list2)
data
```

```python
# Ya tenemos nuestros datos convertidos a Series
print(data)
```

* **Una base de datos es un conjunto de Series.**
* Podemos convertir un Numpy y diccionarios a Series también.

### De `NumPy array` a `Series`:

```python
# Veamos un ejemplo de un vector de Numpy, que es otra estructura de datos muy utilizada en Python
# La clase pasada vimos que un vector es una lista de números:
vector1 = np.array([10, 20, 1, 2, 
                    3, 4, 5, 6, 7, 10])
vector1
```

```python
# Ahora convertimos el vector de Numpy a una Serie de Pandas:
series1 = pd.Series( vector1 )
series1
```

```python
# Las Series de Pandas son más complejas que los vectores de Numpy, ya que permiten almacenar información adicional como el índice y el nombre de la Serie.
print(type(vector1))
print(type(series1))
```

### De `Dictionary` a `Series`:

```python
population_dict= { 'California' : 38332521,
                    'Texas'      : 26448193,
                    'New York'   : 19651127,
                    'Florida'    : 19552860,
                    'Illinois'   : 12882135 }
```

```python
# Vemos las llaves:
population_dict.keys()
```

```python
# Al convertirlo en Series, la indexación absorde las llaves.
# Por tanto, la indexación también puede tener texto. Sin embargo, no es lo ideal. Se recomienda números partiendo desde 0.
population = pd.Series( population_dict )
population
```

## <a id='1.3.'>1.3. `Series` vs `NumPy`</a>

* La diferencia esencial es la presencia del índice: mientras que el `Numpy Array` tiene un índice entero definido de forma implícita que se utiliza para acceder a los valores, los `Pandas Series` tienen un índice explícito asociado a los valores.

* El `index` (índice) no necesita ser un entero. Podemos usar `strings` (cadenas de texto).

```python
# Creamos un vector de Numpy con valores del 5 al 20, con un paso de 2:
claudia = np.arange(5, 21, 2)
claudia
```

```python
# Convertimos el vector de Numpy a una Serie de Pandas, con un índice personalizado:
# Ahora la serie es de 3 en 3, y el índice es un nombre de persona.
math = pd.Series( np.arange(5.,21.,3.) , ['joyce','jeremy','ivan','marcy','daniel','franclin'])
math
```

```python
# Creamos una serie del 5 al 15, de 3 en 3, con un índice personalizado:
info = np.arange(5 , 15 ,3.)
index_info =  ['joyce','jeremy','ivan','marcy']
```

```python
# De los objetos creados anteriormente, podemos crear una Serie de Pandas con un índice personalizado y un nombre para la Serie:
# Students sería el nombre de la Serie, y el índice es el nombre de cada estudiante.
math_2 = pd.Series(  info , index_info,  name = "Students")
math_2
```

# <a id='2.'>II. `DataFrames`</a>

* Un `DataFrame` es una colección de múltiples Series. Puede pensarse como un arreglo bidimensional (2D), donde cada fila es un punto de datos (datapoint) independiente y cada columna es una característica (feature) de los datos. Las filas están etiquetadas con un índice (al igual que en una Series) y las columnas están etiquetadas en el atributo columns.

* Existen muchas formas diferentes de crear `DataFrames`: a partir de listas, diccionarios, etc.

* Usaremos `Pandas` para trabajar con `DataFrames`.


## <a id='2.1.'>2.1. Generación de `DataFrames`</a>
### De `lists` y `dict` a `DataFrame`:

```python
# Notas de estudiantes (lista)
students = [ "Alejandro", "Pedro", "Ramiro", "Axel", "Juan" ]
math     = [ 15, 16, 10, 12, 13 ]
english  = [ 13, 9, 16, 14, 17 ]
art      = [ 12, 16, 15, 19, 10 ]

# Diccionario:
# A cada llave le asignamos una lista 
gradesA = {'Students':students, 'Math':math, 'English':english, 'Art':art}
```

```python
gradesA
```

```python
# Convertimos el diccionario en un DataFrame de Pandas:
gradesA1 = pd.DataFrame(gradesA)
gradesA1
```

## <a id='2.2.'>2.2. Indexing</a>

* Podemos utilizar los mismos métodos que en las `Series`: `iloc` y `loc`, seleccionando tanto columnas como filas.

```python
# Otras notas de estudiantes (lista)
students = [ "Gissela", "Daniel", "Andres", "Sandra", "Rosalyn" ]
math     = [ 16, 14, 17, 17, 17 ]
english  = [ 16, 17, 19, 18, 15 ]
art      = [ 11, 17, 13, 14, 17 ]

# Diccionario
diplomado = {'Students':students, 'Math':math, 'English':english, 'Art':art}

# DataFrame:
gradesA1 = pd.DataFrame( diplomado )
gradesA1
```

```python
# MÉTODO loc(): Con valores de indexación, obtenemos el valor de la Serie.
# En este caso, seleccionamos las filas que van desde el 0 hasta el 3 y las columnas de "Art" y "Students"
# Recordar que loc es un método porque está al costado del objeto y no al costado del alias de la librería (pd como Pandas, por ejemplo).
gradesA1.loc[ 0:3 ,  ["Art", "Students"] ]
```

```python
# Ahora seleccionamos todas las filas (:) y las columnas "Students", "English" y "Art".
gradesA1.loc[ :, ["Students", "English", "Art"] ]
```

## <a id='2.3.'>2.3. Métodos Generales</a>

|Métodos|Descripción|

|dropna( ) |Elimina las entradas nulas.|
|fillna( ) |Reemplaza las entradas nulas con un valor o estrategia especificada.|
|reset_index( )| Convierte el índice en una columna.|

```python
# Veamos otro ejemplo, ahora con diccionarios:
deps = {
        'dep' : ['Lima', 'Piura', 'Tumbes', 'Cuzco', 'Ica', 'Puno'],
        'year': [ 2000, 2001, 2002, 2001, 2002, 2003 ],
        'pop' : [ 1.5, 1.7, 3.6, 2.4, 2.9, 3.2 ] 
        }
dep1 = pd.DataFrame( deps )
dep1
```

* El método `sort_values` tiene muchos argumentos: `inlpace` es uno de ellos y por default es `False`.
* Por ello es importante explicitar cuando lo queremos `True`, para que los cambios tengan efecto en la data ahí mismo. 
* `inplace` = `True` quiere decir que haga efectivo los cambios ahí mismo y que chanques la data. Es razonable para no crear más objetos.

```python
# Quiero ordenarlo según la variable "dep" (departamento): usamos el método sort_values para ordenar.
# Se ordena alfabéticamente porque dep es chr, si fuera variable int se ordena de menor a mayor.
dep1.sort_values( ["dep"] , inplace=True, )
dep1
```

```python
# ascending=False: ordena de mayor a menor, y en letras, de la Z a la A.
dep1.sort_values( ["dep"], inplace= True, ignore_index=True, ascending=False)
dep1
```

## <a id='2.4.'>2.4. Importando datos: GitHub</a>

**INNOVATION GRAPH:**
* Contiene los repositorios por temática y país alrededor del mundo.

```python
# URL del archivo CSV hospedado en GitHub (hacemos click en "Raw")
# Repositorios de Github de la página de Innovation Graph:
url = "https://raw.githubusercontent.com/github/innovationgraph/refs/heads/main/data/repositories.csv"

# Carga directa desde la URL
df = pd.read_csv(url)

# Muestra las primeras filas del DataFrame
df
```

```python
# Renombrar columnas:
df.rename(columns={"repositories": "num_repos"}, inplace=True)
df
```

```python
# Seleccionar solo Perú:
# La idea es seleccionar la columna y colocar los valores que quiero encontrar.
# Para eso se crearon booleanos donde se seleccionarán los valores TRUE
# Traducción: En la base df, quiero los valores del df que son PE
df[df["iso2_code"] == "PE"]
```

```python
# Si quiero dos países: Perú y Argentina
# Usamos un "o" (or) y paréntesis para cada condición
df[ (df["iso2_code"] == "PE") | (df["iso2_code"] == "AR") ]
```

```python
# Quiero los códigos de los países de Sudamérica, que son: PE, AR, BR, CL, CO, EC, GY, PY, SR, UY y VE.
# Usamos el método isin() para seleccionar los valores que están en la lista de países sud
# Creamos lista:
south_america_codes = [
    "AR", "BO", "BR", "CL", "CO", "EC",
    "GY", "PY", "PE", "SR", "UY", "VE", "GF"
]

# Filter DataFrame
df_south_america = df[df["iso2_code"].isin(south_america_codes)]
df_south_america
```

```python
# Queremos saber cuántas columnas y filas hay en DataFrame de países de Sudamérica:
df_south_america.shape
```

```python
# Podemos extraer columnas específicas
# Van dos corchetes porque estamos extrayendo varias columnas, y no una sola. Si fuera una sola columna, bastaría con un solo corchete.
# No chancamos la información, solo nos muestra el resultado.
# El corchete indica que es una lista.
df_south_america[["quarter", "year"]]
```

```python
# Eso implica que podemos reordenar las columnas:
# Con el siglo igual, ahora sí chanco el df
df_south_america = df_south_america[["iso2_code", "year", "quarter", "num_repos"]]
df_south_america
```

```python
# El index sigue siendo caótico, desordenado. Procedemos a ordenarlo y reiniciar el índice:
df_south_america.reset_index(inplace=True)
df_south_america 
```

```python
# Podemos eliminar o mantener la nueva columna index creada
# Decidimos eliminarla porque no es necesaria y solo ocupa espacio.
df_south_america.drop(columns=["index"], inplace=True)
df_south_america
```

```python
# Queremos ordenar el DataFrame por las columnas "iso2_code", "year" y "quarter" en orden ascendente.
# Como si fuera una serie de tiempo: primero por país, luego por año y finalmente por trimestre.
df_south_america = df_south_america.sort_values(
    by=["iso2_code", "year", "quarter"],
    ascending=[True, True, True]
)
df_south_america
```

* Es importante tener en cuenta que la base debe estar balanceada. 
* Es decir, que todos los quearter tengan dato. Si no sucede, hablar con el encargado. Quizás sea conveniente colocar 0 o NaN.

```python
# Quiero ver los valores nulos (NaN) en la columna "num_repos" del DataFrame df_south_america. 
# Esto me permitirá identificar si hay registros sin información.
df_south_america[df_south_america.num_repos.isna() ]
```

```python
# El índice volvió a molestarnos (porque se reordenó el DataFrame). Procedemos a ordenarlo y reiniciar el índice:
df_south_america.reset_index(inplace=True)
df_south_america
```

```python
# Eliminamos el índice anterior y el que se creó al reiniciar el índice, ya que no es necesario y solo ocupa espacio.
# Columnas level_0 e index:
df_south_america.drop(columns=["level_0", "index"], inplace=True)
df_south_america
```

```python
# Quiero ver el tipo de datos de las columnas:
df_south_america.dtypes
```

```python
# Podemos cambiar el tipo de datos de iso2_code a string (str) para asegurarnos de que se trate como texto y no como un número. 
# Esto es útil si queremos realizar operaciones de texto o evitar problemas de interpretación de datos.
df_south_america["iso2_code"] = df_south_america["iso2_code"].astype(str)
df_south_america.dtypes
```

```python
# Exportamos a formato csv en la misma carpeta del notebook:
# Guarda el DataFrame en la carpeta 'data'
df_south_america.to_csv("data/south_america_repositorios.csv", index=False)
```
